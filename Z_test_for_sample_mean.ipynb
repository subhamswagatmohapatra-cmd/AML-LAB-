### $z$-Test for a Sample Mean
This notebook demonstrates how to perform a one-sample $z$-test to compare a sample mean against a known population mean when the population standard deviation is known.

```python
import numpy as np
from statsmodels.stats.weightstats import ztest

# Generate some mock sample data
# Let's assume a null hypothesis (H0) that the population mean is 100
np.random.seed(42)
sample_data = np.random.normal(loc=102, scale=15, size=100)

# Hypothesized population mean
null_mean = 100

# Perform the one-sample z-test
z_stat, p_value = ztest(sample_data, value=null_mean)

print(f"Sample Mean: {np.mean(sample_data):.4f}")
print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print(f"Reject the null hypothesis (p < {alpha}). The sample mean is significantly different from {null_mean}.")
else:
    print(f"Fail to reject the null hypothesis (p >= {alpha}). There is no significant difference from {null_mean}.")
```
