```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
```

```python
from scipy import stats

# Generate sample data for three groups
group1 = np.random.normal(loc=10, scale=2, size=30)
group2 = np.random.normal(loc=12, scale=2, size=30)
group3 = np.random.normal(loc=10, scale=2, size=30)

# Perform one-way ANOVA test
f_statistic, p_value = stats.f_oneway(group1, group2, group3)

print(f"F-statistic: {f_statistic:.2f}")
print(f"P-value: {p_value:.3f}")

# Interpret the results
alpha = 0.05
if p_value < alpha:
    print("\nSince the p-value is less than the significance level (alpha=0.05), we reject the null hypothesis.\nThere is a statistically significant difference between the means of the groups.")
else:
    print("\nSince the p-value is greater than the significance level (alpha=0.05), we fail to reject the null hypothesis.\nThere is no statistically significant difference between the means of the groups.")
```

```python
import pandas as pd
from scipy.stats import spearmanr

# Input data
X = [10, 20, 30, 40, 50, 60]
Y = [15, 25, 35, 30, 45, 55]

# Calculate Spearman Rank Correlation
correlation, p_value = spearmanr(X, Y)

# Display result
print("Spearman Rank Correlation Coefficient:", correlation)
print("P-value:", p_value)

# Interpretation
if correlation > 0:
    print("There is a positive correlation.")
elif correlation < 0:
    print("There is a negative correlation.")
else:
    print("There is no correlation.")
```
