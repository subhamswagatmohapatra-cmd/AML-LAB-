# T-test for a Sample Mean (One-Sample T-test)

A one-sample t-test is used to determine whether the sample mean is statistically different from a known or hypothesized population mean.

```python
import numpy as np
from scipy import stats

# Generate some dummy sample data (e.g., heights of 30 people)
# Suppose the true population mean height is hypothesized to be 170 cm
np.random.seed(42)
sample_data = np.random.normal(loc=172.5, scale=5, size=30)

# Hypothesized population mean
null_mean = 170.0

# Perform the one-sample t-test
t_statistic, p_value = stats.ttest_1samp(sample_data, null_mean)

print(f"Sample Mean: {np.mean(sample_data):.2f}")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
# Interpret the results
alpha = 0.05
if p_value < alpha:
    print(f"Reject the null hypothesis (p < {alpha}). The sample mean is significantly different from {null_mean}.")
else:
    print(f"Fail to reject the null hypothesis (p >= {alpha}). There is no significant difference between the sample mean and {null_mean}.")
```
