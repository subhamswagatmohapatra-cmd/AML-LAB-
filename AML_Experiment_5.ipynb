```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
x=[5,9,13,17,21]
y=[12,20,25,33,35]
def karl_pearson_coefficient(x,y):
  x=np.array(x)
  y=np.array(y)
  correlation_matrix=np.corrcoef(x,y)
  correlation_xy=correlation_matrix[0,1]
  return correlation_xy
coefficient =karl_pearson_coefficient(x,y)
print(f"Karl Pearson Correlation Coefficient:{coefficient:.4f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency, chi2

observed_data=np.array(
    [
        [40,10],
        [25,25]
    ]
)

def run_chi_square_test(data):
  chi2_stat,p,dof,expected=chi2_contingency(data)
  print(f"Chi-Square Statistic:{chi2_stat:.4f}")
  print(f"P-Value:{p:.4f}")
  print(f"Degrees of Freedom:{dof}")
  print("Expected Frequencies:")
  print(expected)

  alpha=0.05
  critical_value = chi2.ppf(1 - alpha, dof)
  print(f"Significance Level (alpha):{alpha:.2f}")
  print(f"Critical Value (from Chi-Square table):{critical_value:.4f}")

  if chi2_stat > critical_value:
    print("Reject the null hypothesis. There is a significant association between the variables.")
  else:
    print("Fail to reject the null hypothesis. There is no significant association between the variables.")

run_chi_square_test(observed_data)
```
